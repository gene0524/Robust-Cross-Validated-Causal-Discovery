# Generate ground truth matrices from the given structure files

```python
import numpy as np
import pandas as pd
import os

def create_adjacency_matrices(input_file, num_nodes):
    """
    Create adjacency matrices from structure file.
    
    Args:
        input_file: Path to input CSV file containing structure information
        num_nodes: Number of nodes in the network
    Returns:
        Tuple of (adjacency matrices list, maximum lag)
    """
    df = pd.read_csv(input_file, names=['cause', 'effect', 'lag'])
    max_lag = df['lag'].max()
    
    # Initialize adjacency matrices
    B_matrices = [np.zeros((num_nodes, num_nodes)) for _ in range(max_lag + 1)]
    
    # Populate the adjacency matrices
    for _, row in df.iterrows():
        cause, effect, lag = row['cause'], row['effect'], row['lag']
        B_matrices[lag][effect, cause] = 1
    
    # Add self-effects with delay=1
    for i in range(num_nodes):
        B_matrices[1][i, i] = 1
        
    print(f"Processing {input_file}")
    print(f"Max lag: {max_lag}")
    print(f"B_matrices shape: {B_matrices[0].shape}")
    
    return B_matrices, max_lag

def save_ground_truth(B_matrices, filepath):
    """
    Save adjacency matrices to file.
    
    Args:
        B_matrices: List of adjacency matrices
        filepath: Output file path
    """
    # Create output directory if it doesn't exist
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    
    with open(filepath, 'w') as f:
        for i, B in enumerate(B_matrices):
            np.savetxt(f, B, delimiter=',', fmt='%.3f')
            if i < len(B_matrices) - 1:
                f.write('\n')

def process_files(file_pairs, num_nodes=25):
    """
    Process multiple input-output file pairs.
    
    Args:
        file_pairs: List of (input_path, output_path) tuples
        num_nodes: Number of nodes in the network (default: 25)
    """
    for input_file, output_file in file_pairs:
        try:
            B_matrices, _ = create_adjacency_matrices(input_file, num_nodes)
            save_ground_truth(B_matrices, output_file)
            print(f"Saved adjacency matrices to {output_file}\n")
        except Exception as e:
            print(f"Error processing {input_file}: {str(e)}\n")
```

```python
file_pairs = [
    ('structures/random-rels_20_1A.csv',
        'ground_truths/random-rels_20_1A_adj.csv'),
    ('structures/random-rels_20_1B.csv',
        'ground_truths/random-rels_20_1B_adj.csv'),
    ('structures/random-rels_20_1C.csv',
        'ground_truths/random-rels_20_1C_adj.csv'),
    ('structures/random-rels_20_1D.csv',
        'ground_truths/random-rels_20_1D_adj.csv'),
    ('structures/random-rels_20_1E.csv',
        'ground_truths/random-rels_20_1E_adj.csv'),
    ('structures/random-rels_20_1_3.csv',
        'ground_truths/random-rels_20_1_3_adj.csv'),
    ('structures/random-rels_40_1_3.csv',
        'ground_truths/random-rels_40_1_3_adj.csv'),
    ('structures/random-rels_40_1.csv',
        'ground_truths/random-rels_40_1_adj.csv'),
    ('structures/manyinputs.csv',
        'ground_truths/manyinputs_adj.csv'),
]

# Process all file pairs
process_files(file_pairs)

print("All files processed successfully.")
```
