# Generate ground truth matrices from the given structure files

```python
import numpy as np
import pandas as pd
import os

def generate_ground_truth_from_structure(dataset_path, structure_path, output_path):
    """
    Generate ground truth matrix from dataset and structure file.
    
    Args:
        dataset_path: Path to dataset CSV file
        structure_path: Path to structure text file
        output_path: Path to save ground truth matrix
    """
    try:
        # Read column names from dataset
        df = pd.read_csv(dataset_path)
        names = df.columns.tolist()
        
        # Remove timestamp if present
        if 'timestamp' in names:
            names.remove('timestamp')
        
        # Initialize matrix with 1s on diagonal
        n = len(names)
        matrix = np.eye(n)
        
        # Read structure file and create relationships
        with open(structure_path, 'r') as f:
            for line in f:
                cause, effect = line.strip().split()
                if cause in names and effect in names:
                    cause_idx = names.index(cause)
                    effect_idx = names.index(effect)
                    matrix[effect_idx][cause_idx] = 1
        
        # Create output directory if it doesn't exist
        os.makedirs(os.path.dirname(output_path), exist_ok=True)
        
        # Save to CSV
        np.savetxt(output_path, matrix, delimiter=',', fmt='%d')
        print(f"Generated ground truth matrix for {os.path.basename(dataset_path)}")
        
        return True
        
    except Exception as e:
        print(f"Error processing {dataset_path}: {str(e)}")
        return False

def process_files(file_triplets):
    """
    Process multiple sets of files to generate ground truths.
    
    Args:
        file_triplets: List of (dataset_path, structure_path, output_path) tuples
    """
    for dataset_path, structure_path, output_path in file_triplets:
        generate_ground_truth_from_structure(dataset_path, structure_path, output_path)
```

```python
# Define activities
activities = [
    "Antivirus_Activity",
    "Middleware_oriented_message_Activity",
    "Web_Activity"
]

# Create file triplets
file_triplets = [
    (f"{activity}/dataset_0.csv",
        f"{activity}/structure.txt",
        f"{activity}/ground_truth.csv")
    for activity in activities
]

# Process all file triplets
process_files(file_triplets)

print("Ground truth generation completed.")
```
